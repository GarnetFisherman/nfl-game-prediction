```python
completed_games = games.filter(
    games["home_score"].is_not_null() &
    games["away_score"].is_not_null()
)
```

```python
completed_games.shape
```

```python
ties = completed_games.filter(
    completed_games["home_score"] == completed_games["away_score"]
)

ties.shape
```

```python
model_games = completed_games.filter(
    completed_games["home_score"] != completed_games["away_score"]
)
```

```python
model_games.shape
```

```python
import polars as pl
```

```python
model_games = model_games.with_columns(
    (pl.col("home_score") > pl.col("away_score"))
    .cast(pl.Int32)
    .alias("home_win")
)
```

```python
model_games.select(
    "away_team",
    "away_score",
    "home_team",
    "home_score",
    "home_win"
).head(10)
```

```python
games_pd = model_games.to_pandas()
```

```python
games_pd[
    (games_pd["home_team"] == "BUF") |
    (games_pd["away_team"] == "BUF")
][
    [
        "season",
        "week",
        "gameday",
        "away_team",
        "away_score",
        "home_team",
        "home_score"
    ]
].tail(20)
```

```python
games_pd = games_pd.sort_values(
    ["season", "week"]
).reset_index(drop=True)
```

```python
games_pd[
    (games_pd["home_team"] == "BUF") |
    (games_pd["away_team"] == "BUF")
][
    [
        "season",
        "week",
        "gameday",
        "away_team",
        "away_score",
        "home_team",
        "home_score"
    ]
].tail(20)
```

```python
home_games = games_pd[
    [
        "game_id",
        "season",
        "week",
        "gameday",
        "home_team",
        "away_team",
        "home_score",
        "away_score"
    ]
].copy()

home_games = home_games.rename(
    columns={
        "home_team": "team",
        "away_team": "opponent",
        "home_score": "points_scored",
        "away_score": "points_allowed"
    }
)

home_games["is_home"] = 1
```

```python
home_games.head()
```

```python
away_games = games_pd[
    [
        "game_id",
        "season",
        "week",
        "gameday",
        "away_team",
        "home_team",
        "away_score",
        "home_score"
    ]
].copy()

away_games = away_games.rename(
    columns={
        "away_team": "team",
        "home_team": "opponent",
        "away_score": "points_scored",
        "home_score": "points_allowed"
    }
)

away_games["is_home"] = 0
```

```python
away_games.head()
```

```python
team_games = pd.concat(
    [home_games, away_games],
    ignore_index=True
)
```

```python
team_games.shape
```

```python
team_games = team_games.sort_values(
    ["team", "season", "week"]
).reset_index(drop=True)
```

```python
team_games[
    team_games["team"] == "BUF"
][
    [
        "season",
        "week",
        "gameday",
        "team",
        "opponent",
        "points_scored",
        "points_allowed",
        "is_home"
    ]
].tail(20)
```

```python
team_games["prev_points_scored"] = (
    team_games.groupby("team")["points_scored"]
    .shift(1)
)
```

```python
team_games[
    team_games["team"] == "BUF"
][
    [
        "season",
        "week",
        "opponent",
        "points_scored",
        "prev_points_scored"
    ]
].head(10)
```

```python
team_games["avg_points_scored_last5"] = (
    team_games.groupby("team")["points_scored"]
    .transform(lambda x: x.shift(1).rolling(5).mean())
)
```

```python
team_games[
    team_games["team"] == "BUF"
][
    [
        "season",
        "week",
        "opponent",
        "points_scored",
        "prev_points_scored",
        "avg_points_scored_last5"
    ]
].head(12)
```

```python
team_games["avg_points_allowed_last5"] = (
    team_games.groupby("team")["points_allowed"]
    .transform(lambda x: x.shift(1).rolling(5).mean())
)
```

```python
team_games[
    team_games["team"] == "BUF"
][
    [
        "season",
        "week",
        "opponent",
        "points_scored",
        "points_allowed",
        "avg_points_scored_last5",
        "avg_points_allowed_last5"
    ]
].head(12)
```

```python
team_games["avg_point_diff_last5"] = (
    team_games["avg_points_scored_last5"]
    - team_games["avg_points_allowed_last5"]
)
```

```python
team_games[
    team_games["team"] == "BUF"
][
    [
        "season",
        "week",
        "opponent",
        "points_scored",
        "points_allowed",
        "avg_points_scored_last5",
        "avg_points_allowed_last5",
        "avg_point_diff_last5"
    ]
].head(12)
```

```python
team_games["team_win"] = (
    team_games["points_scored"] > team_games["points_allowed"]
).astype(int)
```

```python
team_games[
    team_games["team"] == "BUF"
][
    [
        "season",
        "week",
        "opponent",
        "points_scored",
        "points_allowed",
        "team_win"
    ]
].head(10)
```

```python
team_games["win_pct_last5"] = (
    team_games.groupby("team")["team_win"]
    .transform(lambda x: x.shift(1).rolling(5).mean())
)
```

```python
team_games[
    team_games["team"] == "BUF"
][
    [
        "season",
        "week",
        "opponent",
        "points_scored",
        "team_win",
        "avg_points_scored_last5",
        "avg_points_allowed_last5",
        "avg_point_diff_last5",
        "win_pct_last5"
    ]
].head(12)
```
