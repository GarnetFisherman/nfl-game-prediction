```python
test_seasons = [2022, 2023, 2024, 2025]
```

```python
walk_forward_results = []
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, log_loss, brier_score_loss

walk_forward_results = []

for test_season in test_seasons:

    # Train only on seasons before the test season
    season_train = model_ready[
        model_ready["season"] < test_season
    ].copy()

    # Test only on the selected season
    season_test = model_ready[
        model_ready["season"] == test_season
    ].copy()

    # Separate features and targets
    X_train_season = season_train[feature_columns]
    y_train_season = season_train["home_win"]

    X_test_season = season_test[feature_columns]
    y_test_season = season_test["home_win"]

    # Create and train a new model for this test season
    season_model = LogisticRegression()
    season_model.fit(X_train_season, y_train_season)

    # Generate classifications and probabilities
    season_predictions = season_model.predict(X_test_season)
    season_probabilities = season_model.predict_proba(
        X_test_season
    )[:, 1]

    # Always-home accuracy baseline
    home_baseline_accuracy = y_test_season.mean()

    # Store the season's results
    walk_forward_results.append({
        "test_season": test_season,
        "training_games": len(season_train),
        "test_games": len(season_test),
        "model_accuracy": accuracy_score(
            y_test_season,
            season_predictions
        ),
        "home_baseline_accuracy": home_baseline_accuracy,
        "log_loss": log_loss(
            y_test_season,
            season_probabilities
        ),
        "brier_score": brier_score_loss(
            y_test_season,
            season_probabilities
        )
    })
```

```python
walk_forward_df = pd.DataFrame(walk_forward_results)

walk_forward_df
```

```python
walk_forward_df["accuracy_improvement"] = (
    walk_forward_df["model_accuracy"]
    - walk_forward_df["home_baseline_accuracy"]
)

walk_forward_df
```

```python
summary_metrics = pd.Series({
    "mean_model_accuracy": walk_forward_df["model_accuracy"].mean(),
    "mean_home_baseline_accuracy": walk_forward_df[
        "home_baseline_accuracy"
    ].mean(),
    "mean_accuracy_improvement": walk_forward_df[
        "accuracy_improvement"
    ].mean(),
    "mean_log_loss": walk_forward_df["log_loss"].mean(),
    "mean_brier_score": walk_forward_df["brier_score"].mean()
})

summary_metrics
```

```python
def add_rolling_team_features(df, window):
    result = df.copy()

    result[f"avg_points_scored_last{window}"] = (
        result.groupby("team")["points_scored"]
        .transform(
            lambda x: x.shift(1).rolling(window).mean()
        )
    )

    result[f"avg_points_allowed_last{window}"] = (
        result.groupby("team")["points_allowed"]
        .transform(
            lambda x: x.shift(1).rolling(window).mean()
        )
    )

    result[f"avg_point_diff_last{window}"] = (
        result[f"avg_points_scored_last{window}"]
        - result[f"avg_points_allowed_last{window}"]
    )

    result[f"win_pct_last{window}"] = (
        result.groupby("team")["team_win"]
        .transform(
            lambda x: x.shift(1).rolling(window).mean()
        )
    )

    return result
```

```python
def add_rolling_epa_features(df, window):
    result = df.copy()

    result[f"off_epa_last{window}"] = (
        result.groupby("team")["off_epa_per_play"]
        .transform(
            lambda x: x.shift(1).rolling(window).mean()
        )
    )

    result[f"def_epa_last{window}"] = (
        result.groupby("team")["def_epa_per_play"]
        .transform(
            lambda x: x.shift(1).rolling(window).mean()
        )
    )

    return result
```

```python
def build_model_dataset(window):

    # -------------------------
    # Rolling team stats
    # -------------------------
    tg = add_rolling_team_features(
        team_games,
        window
    )

    # Home team features
    home = tg[tg["is_home"] == 1][
        [
            "game_id",
            f"avg_points_scored_last{window}",
            f"avg_points_allowed_last{window}",
            f"avg_point_diff_last{window}",
            f"win_pct_last{window}"
        ]
    ].copy()

    home = home.rename(columns={
        f"avg_points_scored_last{window}":
            "home_avg_points_scored",
        f"avg_points_allowed_last{window}":
            "home_avg_points_allowed",
        f"avg_point_diff_last{window}":
            "home_avg_point_diff",
        f"win_pct_last{window}":
            "home_win_pct"
    })

    # Away team features
    away = tg[tg["is_home"] == 0][
        [
            "game_id",
            f"avg_points_scored_last{window}",
            f"avg_points_allowed_last{window}",
            f"avg_point_diff_last{window}",
            f"win_pct_last{window}"
        ]
    ].copy()

    away = away.rename(columns={
        f"avg_points_scored_last{window}":
            "away_avg_points_scored",
        f"avg_points_allowed_last{window}":
            "away_avg_points_allowed",
        f"avg_point_diff_last{window}":
            "away_avg_point_diff",
        f"win_pct_last{window}":
            "away_win_pct"
    })

    # -------------------------
    # Rolling EPA
    # -------------------------
    te = add_rolling_epa_features(
        team_epa,
        window
    )

    # Home EPA
    home_epa_temp = te[
        [
            "game_id",
            "team",
            f"off_epa_last{window}",
            f"def_epa_last{window}"
        ]
    ].merge(
        games_pd[
            ["game_id", "home_team"]
        ],
        on="game_id",
        how="inner"
    )

    home_epa_temp = home_epa_temp[
        home_epa_temp["team"]
        == home_epa_temp["home_team"]
    ].copy()

    home_epa_temp = home_epa_temp.rename(columns={
        f"off_epa_last{window}": "home_off_epa",
        f"def_epa_last{window}": "home_def_epa"
    })

    home_epa_temp = home_epa_temp[
        [
            "game_id",
            "home_off_epa",
            "home_def_epa"
        ]
    ]

    # Away EPA
    away_epa_temp = te[
        [
            "game_id",
            "team",
            f"off_epa_last{window}",
            f"def_epa_last{window}"
        ]
    ].merge(
        games_pd[
            ["game_id", "away_team"]
        ],
        on="game_id",
        how="inner"
    )

    away_epa_temp = away_epa_temp[
        away_epa_temp["team"]
        == away_epa_temp["away_team"]
    ].copy()

    away_epa_temp = away_epa_temp.rename(columns={
        f"off_epa_last{window}": "away_off_epa",
        f"def_epa_last{window}": "away_def_epa"
    })

    away_epa_temp = away_epa_temp[
        [
            "game_id",
            "away_off_epa",
            "away_def_epa"
        ]
    ]

    # -------------------------
    # Merge everything
    # -------------------------
    result = games_pd.merge(
        home,
        on="game_id",
        how="left"
    )

    result = result.merge(
        away,
        on="game_id",
        how="left"
    )

    result = result.merge(
        home_epa_temp,
        on="game_id",
        how="left"
    )

    result = result.merge(
        away_epa_temp,
        on="game_id",
        how="left"
    )

    # -------------------------
    # Matchup differences
    # Positive = home advantage
    # -------------------------
    result["points_scored_diff"] = (
        result["home_avg_points_scored"]
        - result["away_avg_points_scored"]
    )

    result["point_diff_diff"] = (
        result["home_avg_point_diff"]
        - result["away_avg_point_diff"]
    )

    result["win_pct_diff"] = (
        result["home_win_pct"]
        - result["away_win_pct"]
    )

    result["off_epa_diff"] = (
        result["home_off_epa"]
        - result["away_off_epa"]
    )

    # Lower defensive EPA allowed is better,
    # so reverse the subtraction.
    result["def_epa_advantage"] = (
        result["away_def_epa"]
        - result["home_def_epa"]
    )

    result["rolling_window"] = window

    return result
```

```python
def evaluate_window(df, window, features, test_seasons):
    results = []

    for test_season in test_seasons:

        train = df[
            df["season"] < test_season
        ].copy()

        test = df[
            df["season"] == test_season
        ].copy()

        X_train = train[features]
        y_train = train["home_win"]

        X_test = test[features]
        y_test = test["home_win"]

        model = LogisticRegression(
            max_iter=1000
        )

        model.fit(X_train, y_train)

        predictions = model.predict(X_test)

        probabilities = model.predict_proba(
            X_test
        )[:, 1]

        results.append({
            "window": window,
            "test_season": test_season,
            "training_games": len(train),
            "test_games": len(test),
            "accuracy": accuracy_score(
                y_test,
                predictions
            ),
            "log_loss": log_loss(
                y_test,
                probabilities
            ),
            "brier_score": brier_score_loss(
                y_test,
                probabilities
            )
        })

    return pd.DataFrame(results)
```

```python
def evaluate_scaled_logistic(df, features, test_seasons):
    results = []

    for test_season in test_seasons:

        train = df[
            df["season"] < test_season
        ].copy()

        test = df[
            df["season"] == test_season
        ].copy()

        X_train = train[features]
        y_train = train["home_win"]

        X_test = test[features]
        y_test = test["home_win"]

        model = Pipeline([
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(max_iter=1000))
        ])

        model.fit(X_train, y_train)

        predictions = model.predict(X_test)
        probabilities = model.predict_proba(X_test)[:, 1]

        results.append({
            "test_season": test_season,
            "accuracy": accuracy_score(y_test, predictions),
            "log_loss": log_loss(y_test, probabilities),
            "brier_score": brier_score_loss(
                y_test,
                probabilities
            )
        })

    return pd.DataFrame(results)
```

```python
def evaluate_calibration_methods(
    df,
    features,
    test_seasons
):
    results = []
    predictions = []

    for test_season in test_seasons:

        train = df[
            df["season"] < test_season
        ].sort_values(
            ["season", "week"]
        ).copy()

        test = df[
            df["season"] == test_season
        ].sort_values(
            ["season", "week"]
        ).copy()

        X_train = train[features]
        y_train = train["home_win"]

        X_test = test[features]
        y_test = test["home_win"]

        methods = {
            "raw": Pipeline([
                ("scaler", StandardScaler()),
                (
                    "model",
                    LogisticRegression(max_iter=1000)
                )
            ]),

            "sigmoid": CalibratedClassifierCV(
                estimator=Pipeline([
                    ("scaler", StandardScaler()),
                    (
                        "model",
                        LogisticRegression(max_iter=1000)
                    )
                ]),
                method="sigmoid",
                cv=TimeSeriesSplit(n_splits=3),
                ensemble=False
            ),

            "isotonic": CalibratedClassifierCV(
                estimator=Pipeline([
                    ("scaler", StandardScaler()),
                    (
                        "model",
                        LogisticRegression(max_iter=1000)
                    )
                ]),
                method="isotonic",
                cv=TimeSeriesSplit(n_splits=3),
                ensemble=False
            )
        }

        for method_name, model in methods.items():

            model.fit(X_train, y_train)

            probabilities = model.predict_proba(
                X_test
            )[:, 1]

            classifications = (
                probabilities >= 0.5
            ).astype(int)

            results.append({
                "test_season": test_season,
                "method": method_name,
                "games": len(test),
                "accuracy": accuracy_score(
                    y_test,
                    classifications
                ),
                "log_loss": log_loss(
                    y_test,
                    probabilities
                ),
                "brier_score": brier_score_loss(
                    y_test,
                    probabilities
                )
            })

            season_predictions = pd.DataFrame({
                "test_season": test_season,
                "method": method_name,
                "actual": y_test.values,
                "home_win_probability": probabilities
            })

            predictions.append(season_predictions)

    return (
        pd.DataFrame(results),
        pd.concat(
            predictions,
            ignore_index=True
        )
    )
```

```python
def evaluate_sigmoid_calibration(df, features, test_seasons):
    results = []
    predictions = []

    for test_season in test_seasons:

        calibration_season = test_season - 1

        # Base model training uses data before the calibration season
        train = df[
            df["season"] < calibration_season
        ].copy()

        # Most recent historical season is reserved for calibration
        calibrate = df[
            df["season"] == calibration_season
        ].copy()

        # Completely untouched future season
        test = df[
            df["season"] == test_season
        ].copy()

        X_train = train[features]
        y_train = train["home_win"]

        X_cal = calibrate[features]
        y_cal = calibrate["home_win"]

        X_test = test[features]
        y_test = test["home_win"]

        # Base model
        base_model = Pipeline([
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(max_iter=1000))
        ])

        base_model.fit(X_train, y_train)

        # Tell sklearn this model is already fitted
        calibrated_model = CalibratedClassifierCV(
            estimator=base_model,
            method="sigmoid",
            cv="prefit"
        )

        calibrated_model.fit(X_cal, y_cal)

        probabilities = calibrated_model.predict_proba(
            X_test
        )[:, 1]

        classifications = (
            probabilities >= 0.5
        ).astype(int)

        results.append({
            "test_season": test_season,
            "calibration_season": calibration_season,
            "accuracy": accuracy_score(
                y_test,
                classifications
            ),
            "log_loss": log_loss(
                y_test,
                probabilities
            ),
            "brier_score": brier_score_loss(
                y_test,
                probabilities
            )
        })

        predictions.append(
            pd.DataFrame({
                "test_season": test_season,
                "actual": y_test.values,
                "probability": probabilities
            })
        )

    return (
        pd.DataFrame(results),
        pd.concat(predictions, ignore_index=True)
    )
```

```python
def evaluate_sigmoid_calibration(df, features, test_seasons):

    results = []
    predictions = []

    for test_season in test_seasons:

        calibration_season = test_season - 1

        # Older data trains the football model
        train = df[
            df["season"] < calibration_season
        ].copy()

        # Most recent historical season calibrates probabilities
        calibrate = df[
            df["season"] == calibration_season
        ].copy()

        # Future season remains untouched
        test = df[
            df["season"] == test_season
        ].copy()

        # -------------------------
        # Base football model
        # -------------------------

        base_model = Pipeline([
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(max_iter=1000))
        ])

        base_model.fit(
            train[features],
            train["home_win"]
        )

        # Raw probabilities on calibration season
        cal_probs = base_model.predict_proba(
            calibrate[features]
        )[:, 1]

        # -------------------------
        # Sigmoid / Platt calibrator
        # -------------------------

        calibrator = LogisticRegression()

        calibrator.fit(
            cal_probs.reshape(-1, 1),
            calibrate["home_win"]
        )

        # -------------------------
        # Predict untouched season
        # -------------------------

        raw_test_probs = base_model.predict_proba(
            test[features]
        )[:, 1]

        calibrated_probs = calibrator.predict_proba(
            raw_test_probs.reshape(-1, 1)
        )[:, 1]

        classifications = (
            calibrated_probs >= 0.5
        ).astype(int)

        results.append({
            "test_season": test_season,
            "calibration_season": calibration_season,
            "accuracy": accuracy_score(
                test["home_win"],
                classifications
            ),
            "log_loss": log_loss(
                test["home_win"],
                calibrated_probs
            ),
            "brier_score": brier_score_loss(
                test["home_win"],
                calibrated_probs
            )
        })

        predictions.append(
            pd.DataFrame({
                "test_season": test_season,
                "actual": test["home_win"].values,
                "probability": calibrated_probs
            })
        )

    return (
        pd.DataFrame(results),
        pd.concat(
            predictions,
            ignore_index=True
        )
    )
```
