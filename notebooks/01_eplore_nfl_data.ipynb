```python
games.filter(
    games["home_score"].is_null() | games["away_score"].is_null()
).select(
    "season",
    "week",
    "gameday",
    "away_team",
    "away_score",
    "home_team",
    "home_score"
)
```

```python
completed_games = games.filter(
    games["home_score"].is_not_null() &
    games["away_score"].is_not_null()
)
```

```python
completed_games.shape
```

```python
ties = completed_games.filter(
    completed_games["home_score"] == completed_games["away_score"]
)

ties.shape
```

```python
model_games = completed_games.filter(
    completed_games["home_score"] != completed_games["away_score"]
)
```

```python
model_games.shape
```

```python
import polars as pl
```

```python
model_games = model_games.with_columns(
    (pl.col("home_score") > pl.col("away_score"))
    .cast(pl.Int32)
    .alias("home_win")
)
```

```python
model_games.select(
    "away_team",
    "away_score",
    "home_team",
    "home_score",
    "home_win"
).head(10)
```

```python
games_pd = model_games.to_pandas()
```

```python
type(games_pd)
```

```python
games_pd.shape
```

```python
games_pd.info()
```

```python
games_pd["home_win"].value_counts()
```

```python
games_pd["home_win"].value_counts(normalize=True)
```

```python
games_pd.groupby("season")["home_win"].mean()
```

```python
home_win_by_season = games_pd.groupby("season")["home_win"].mean()
```

```python
import matplotlib.pyplot as plt

home_win_by_season.plot()

plt.title("NFL Home Team Win Rate by Season")
plt.xlabel("Season")
plt.ylabel("Home Win Rate")

plt.show()
```

```python
games_pd.groupby("home_team")["home_win"].agg(["count", "mean"])
```

```python
team_home_stats = (
    games_pd.groupby("home_team")["home_win"]
    .agg(["count", "mean"])
    .sort_values("mean", ascending=False)
)

team_home_stats
```
